```python
# 1. reshape --> you can use this to create a new shaoe if the array.
# Suppose we have a 1 D array of 20 elements.
# Now to reshape it, we need to find the factors of 20 : 1,20,4,5,2,10

random_arr = np.random.randint(1,50,(20))
random_arr
```

```python
# Now reshape

# First i will reshape in (1,20)

print("================2-D===================")

print(random_arr.reshape((1,20)))

print("================2-D===================")

print(random_arr.reshape((20,1)))

print("================2-D===================")

print(random_arr.reshape((2,10)))

print("================2-D===================")

print(random_arr.reshape((10,2)))

print("================2-D===================")

print(random_arr.reshape((4,5)))

print("================3-D===================")

print(random_arr.reshape((2,2,5)))

print("================3-D===================")

print(random_arr.reshape((5,2,2)))

print("================3-D===================")

print(random_arr.reshape((10,2,1)))
```
