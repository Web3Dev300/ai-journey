```python
cur_x = 5 # The algorithm starts at  x=3

rate = 0.1 # Learning rate

precision = 0.5 #This tells us when to stop the algorithm
previous_step_size = 1 #
max_iters = 1000000 # Maximum number of iteration
iters = 0 # iteration counter

df = lambda x: 2*(x+5) #Gradient of our function

while previous_step_size > precision and iters < max_iters:
    prev_x = cur_x # Store current x value in prev_x
    cur_x = cur_x - rate * df(prev_x) # Gradient Descent
    previous_step_size = abs(cur_x - prev_x) # Change in x
    iters = iters + 1 # Iteration count
    print("Iteration",iters, "\nX value is", cur_x) #Print iterations
print("Thje local minimum occurs at", cur_x)
```
