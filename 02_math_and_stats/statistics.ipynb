```python
import pandas as pd

df = pd.DataFrame({'Name': ['Jane','Michael','Willian', 'Rosy', 'Hana', 'Ferdie', 'Graeme'], 'Salary':[50000,54000,50000,189000,55000,40000,59000]})

print(df.head())
```

```python
# Calculate the mean, median, mode, min, max

print('Min: ' + str(df['Salary'].min()))
print('Max: ' + str(df['Salary'].max()))
print('Means: ', df['Salary'].mean())
print('Median: ', df['Salary'].median())
print('Mode: ', df['Salary'].mode())
print('Range: ', df['Salary'].max() - df['Salary'].min())
```

```python
df['Salary'].describe()
df['Salary'].describe()['count']
df['Salary'].describe()['mean']
```

```python
# Histogram of salaries

%matplotlib inline
import matplotlib.pyplot as plt

print(df)

salary = df['Salary']
salary.plot.hist(title='Salary Distributioun', color='gray', bins=25)
plt.axvline(salary.mean(), color='violet', linestyle='dashed', linewidth=2)
plt.axvline(salary.median(), color='red', linestyle='dashed', linewidth=2)
plt.show()
```
